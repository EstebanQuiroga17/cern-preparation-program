```python
import numpy as np
import matplotlib.pyplot as plt
```


# 1. Carga de datos


```python
ruta_archivo = "inflammation-01.csv"
datos = np.loadtxt(ruta_archivo, delimiter=",")
```

# 2. Exploracion basica


```python
print("Informacion basica del conjunto de datos")
print(f"Filas (pacientes): {datos.shape[0]}")
print(f"Columnas (dias): {datos.shape[1]}")
print(f"Tipo de datos: {datos.dtype}")
print(f"Valor minimo: {datos.min()}")
print(f"Valor maximo: {datos.max()}")
```

# 3. Analisis: estadisticas por dia y total por paciente

```python
media_por_dia = np.mean(datos, axis=0)
desviacion_por_dia = np.std(datos, axis=0)
total_por_paciente = np.sum(datos, axis=1)

dia_mayor_media = np.argmax(media_por_dia) + 1
paciente_mayor_total = np.argmax(total_por_paciente) + 1

print("\nResultados del analisis")
print(f"Media general: {datos.mean():.2f}")
print(f"Media del primer dia: {media_por_dia[0]:.2f}")
print(f"Media del ultimo dia: {media_por_dia[-1]:.2f}")
print(f"Dia con mayor media de inflamacion: {dia_mayor_media}")
print(f"Total maximo de inflamacion: {total_por_paciente.max():.2f}")
print(f"Paciente con mayor total: {paciente_mayor_total}")
```

# 4. Visualizacion

```python
dias = np.arange(1, datos.shape[1] + 1)

plt.figure(figsize=(10, 5))
plt.plot(dias, media_por_dia, marker="o", color="steelblue", label="Media diaria")
plt.fill_between(
    dias,
    media_por_dia - desviacion_por_dia,
    media_por_dia + desviacion_por_dia,
    color="steelblue",
    alpha=0.2,
    label="Desviacion estandar",
)
plt.title("Evolucion promedio de la inflamacion")
plt.xlabel("Dia")
plt.ylabel("Nivel de inflamacion")
plt.xticks(dias)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

# 5. Interpretacion

    "Los niveles de inflamacion aumentan durante los primeros dias,
    alcanzan sus valores mas altos aproximadamente en la mitad del periodo
    y luego disminuyen hacia el final.
